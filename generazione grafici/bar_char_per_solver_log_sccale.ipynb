```python
import matplotlib.pyplot as plt
import numpy as np

LOG_SCALE = False
# ---------------------------
# Configurazione grafico
# ---------------------------
plt.rcParams.update({
    "font.size": 11,
    "font.family": "serif",
    "axes.labelsize": 12,
    "axes.titlesize": 12,
    "legend.fontsize": 9,
    "figure.dpi": 300
})

# ---------------------------
# ID aggiornati
# ---------------------------
shallow_ids = ["SH-30", "SH-100", "SH-200", "SH-500", "SH-1000", "SH-2000", "SH-4000", "SH-8000"]
complex_ids = ["CX-50", "CX-100", "CX-250", "CX-500"]
conv_ids = ["CV-5", "CV-15", "CV-25", "CV-50", "CV-100", "CV-500"]

# ---------------------------
# DATI
# ---------------------------
# Shallow
shallow_baseline = {
    "Crown": [0, 0, 0, 0, 2, 2, 5, 2],
    "Pyrat": [62, 13, 9, 11, 7, 7, 8, 7],
    "Marabou": [30, 11, 7, 6, 6, 4, 5, 99.9],
    "PyNever": [0, 1, 3, 12, 11, 13, 14, 14]
}

shallow_nsp = {
    "Crown": [0, 0, 0, 0, 0, 0, 0, 0],
    "Pyrat": [62, 14, 16, 9, 5, 5, 3, 0],
    "Marabou": [30, 0, 0, 0, 0, 0, 0, 0],
    "PyNever": [0, 0, 0, 0, 0, 0, 0, 0]
}

# Complex
complex_baseline = {
    "Crown": [2, 9, 12, 14],
    "Pyrat": [56, 63, 57, 61],
    "Marabou": [44, 45, 36, 35],
    "PyNever": [44, 77, 91, 92]
}

complex_nsp = {
    "Crown": [2, 0, 0, 0],
    "Pyrat": [56, 11, 9, 30],
    "Marabou": [44, 6, 3, 18],
    "PyNever": [44, 0, 3, 11]
}

# Convolutional
conv_baseline = {
    "Crown": [24, 1, 0, 0, 1, 3],
    "Pyrat": [77, 19, 24, 22, 63, 54],
    "Marabou": [52, 8, 10, 7, 25, 13]
}

conv_nsp = {
    "Crown": [24, 0, 0, 0, 0, 0],
    "Pyrat": [77, 1, 0, 1, 0, 0],
    "Marabou": [52, 1, 0, 10, 0, 0]
}

# ---------------------------
# PATTERN + COLOR
# ---------------------------
hatches = {"Baseline": "////", "NSP": "...."}
colors = {"Baseline": 'C0', "NSP": 'C1'}  # colori distinti


# ---------------------------
# FUNZIONE GRAFICO
# ---------------------------
def plot_solver_color_pattern_vlines(solver, shallow, complex_, conv, shallow_nsp, complex_nsp, conv_nsp):
    if solver == "PyNever":
        ids = shallow_ids + complex_ids
        baseline = shallow[solver] + complex_[solver]
        nsp = shallow_nsp[solver] + complex_nsp[solver]
    else:
        ids = shallow_ids + complex_ids + conv_ids
        baseline = shallow[solver] + complex_[solver] + conv[solver]
        nsp = shallow_nsp[solver] + complex_nsp[solver] + conv_nsp[solver]

    # Evita problemi con log scale (sostituisce 0 con 0.1)
    baseline = [v if v > 0 else 0.1 for v in baseline]
    nsp = [v if v > 0 else 0.1 for v in nsp]

    x = np.arange(len(ids))
    width = 0.35
    fig, ax = plt.subplots(figsize=(12, 4))

    # Bar plot con hatch
    bars_base = ax.bar(x - width / 2, baseline, width,
                       edgecolor='black', color=colors["Baseline"], hatch=hatches["Baseline"],
                       label='Baseline')
    bars_nsp = ax.bar(x + width / 2, nsp, width,
                      edgecolor='black', color=colors["NSP"], hatch=hatches["NSP"],
                      label='NSP')

    # punti sui valori reali
    for bar in bars_base:
        height = max(bar.get_height(), 0.1)
        ax.scatter(bar.get_x() + bar.get_width() / 2, height, color='black', zorder=10, s=30)
    for bar in bars_nsp:
        height = max(bar.get_height(), 0.1)
        ax.scatter(bar.get_x() + bar.get_width() / 2, height, color='red', zorder=10, s=30)

    # Etichette X
    ax.set_xticks(x)
    ax.set_xticklabels(ids, rotation=45, ha='right', fontsize=10)

    ax.set_title(f"{solver}")
    # Log scale
    if LOG_SCALE:
        ax.set_yscale('log')
        ax.set_ylabel("Timeouts (log scale)")
    else:
        ax.set_ylabel("Timeouts")
    # Linee verticali per separare SH / CX / CV
    sh_end = len(shallow_ids) - 0.5
    cx_end = len(shallow_ids) + len(complex_ids) - 0.5
    ax.axvline(sh_end, color='black', linestyle='-', linewidth=1)
    if solver != "PyNever":
        ax.axvline(cx_end, color='black', linestyle='-', linewidth=1)

    # Legenda in alto a destra
    ax.legend(frameon=False, ncol=2, loc='upper right')

    plt.tight_layout()
    plt.savefig(f"{solver.lower()}_timeouts.jpeg", bbox_inches="tight", format="jpeg")
    plt.close()


# ---------------------------
# GENERA GRAFICI
# ---------------------------
for solver in ["Crown", "Pyrat", "Marabou", "PyNever"]:
    plot_solver_color_pattern_vlines(solver, shallow_baseline, complex_baseline, conv_baseline,
                                     shallow_nsp, complex_nsp, conv_nsp)

print("✅ Grafici aggiornati con linee verticali, hatches e punti sulle barre.")
```
